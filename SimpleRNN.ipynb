```python

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Input sequences
X = np.array([
    [[1], [2], [3]],
    [[2], [3], [4]],
    [[3], [4], [5]],
    [[4], [5], [6]]
], dtype=float)

# Expected next numbers
y = np.array([4, 5, 6, 7], dtype=float)

# Create RNN model
model = Sequential([
    SimpleRNN(8, activation="tanh",
              input_shape=(3, 1)),
    Dense(1)
])

# Compile model
model.compile(optimizer="adam", loss="mse")

# Train model
model.fit(X, y, epochs=200, verbose=0)

# Predict the next number after 5, 6, 7
new_data = np.array([[[5], [6], [7]]], dtype=float)
print(model.predict(new_data, verbose=0))
```
